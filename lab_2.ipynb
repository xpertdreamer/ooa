## Теоретическая часть

### 1. Что называют табличными данными?

Табличные данные — это структурированный набор информации, или данные, которые можно представить в виде таблицы, где строки — это отдельные записи, а столбцы — различные характеристики исследуемого объекта.

### 2. Чем табличные данные отличаются от: изображений; текста; временных рядов?

  * В изображениях важна пространственная структура: пиксели образуют сетку, и соседние точки связаны между собой, из чего складывается картинка. Поэтому для обработки нужны модели, которые учитывают это соседство.
  * Текст представляет собой последовательность слов переменной длины, в которой смысл во многом зависит от порядка и контекста, в отличие от таблиц, где порядок, в целом, не важен.
  * Временные ряды это те же табличные данные, но упорядоченные по времени, где каждое значение зависит от предыдущих. Порядок здесь критичен для прогноза и поиска аномалий, а если значения перемешать — информация потеряется.


### 3. Почему большинство промышленных ML-задач работают именно с табличными данными?

На это есть ряд причин, в центре которых стоит бизнес-логика. Чаще всего компании уже хранят необходимые данные в виде таблиц, то есть реляционных баз данных, а признаки в таких таблицах часто осмыслены и структурированы. Также по таблице легче объяснить поведение модели и встроить в уже существующие системы.

### 4. Какие типы признаков встречаются в табличных данных?

Выделяются 4 основных типа - числовые, категориальные, временные и текстовые. Категориальные и текстовые кодируются по одному из существующих методов, то есть переводятся в числовой формат, а временные можно разбить на несколько признаков, например, день недели, месяц и т.п.

### 5. Что такое признак и целевая переменная?

* Признак - это характеристика исследуемого объекта, используемая как входные данные для совершения моделью прогноза, или индивидуальное измеримое свойство или характеристика наблюдаемого явления. Выбор информативных, отличительных и независимых признаков напрямую влияет на эффективность алгоритма.
* Целевая переменная - индивидуальное измеримое свойство или характеристика наблюдаемого явления или объекта, которое модель машинного обучения пытается предсказать на основе входных данных (признаков), от которых она зависит.

### 6. Различия между: числовыми; категориальными; бинарными признаками.

Числовые признаки принимают значения из непрерывного или дискретного диапазона, в то время как категориальные описывают принадлежность объекта к какой-либо выделяемой группе, часто кодируются в виде целых чисел. Над категориальными проводить арифметические операции не имеет смысла, так как видимого результат данная процедура не имеет. Бинарные признаки в какой-то степени можно назвать частным случаем категориальных данных; они имеют два потенциальных значения - истина/ложь или 0/1 в зависимости от интерпретации; не требуют кодирования, как и числовые.

### 7. Почему корректное кодирование признаков критично для моделей ML?

Преобразования данных — важный шаг в подготовке ваших данных к задачам машинного обучения, так как:
* Алгоритмы машинного обучения ожидают данные в определённом формате, обычно числовом.
* Правильное кодирование данных гарантирует, что алгоритмы могут извлекать значимые закономерности и связи из данных.
* Отсутсвие приведения числовых признаков с большим разбросом к единому масштабу может привести к доминированию признаков с большей величиной.
* Неправильное применение алгоритмов кодирования может привести к значительному росту времени обучения и переобучения на редких категориях.
* Несоглсованность разных типов выборок может привести к неправильной работе модели.

## Анализ

##### Самые многочисленные

Самыми частыми можно считать комбинации (2,1), (2,0), (1,0), (1,1). Вместе они охватывают большую часть датасета, что подтверждает заявленнный дисбаланс.

##### Аномальные и нетипичные

1. Из 14 теоретически возможных комбинаций (7 классов × 2 значения Wilderness_Area1) в датасете присутствуют 11. Отсутствуют три комбинации: (3, 1), (4, 1), (6, 1), то есть классы 3, 4 и 6 не встречаются в первой зоне дикой природы.
2. Малые группы (5, 1), (4, 0) статистически нетипичны из-за малого числа объектов.
3. Комбинация (4, 0) имеет сравнительно низкие значения по всем показателям.

Используя данные из лабораторной:

объедините две таблицы с помощью merge или объедините несколько таблиц с помощью concat.

    

```python
data1 = df.head(50)
cover_type_info = pd.DataFrame({
    'Cover_Type': [1, 2, 3, 4, 5, 6, 7],
    'Avg_Temperature_C': [4.5, 5.2, 8.1, 11.3, 7.8, 9.4, 2.1],
})
df_merged = pd.merge(data1, cover_type_info, on='Cover_Type', how='left')
df_merged
```

```python
print(f"Shape: {df_merged.shape}")
print(df_merged.isna().sum().sum())
```

## Анализ

Размер итоговой таблицы составил 50 строк и 56 столбцов.

Объединение выполнялось по ключу Cover_Type - это общий столбец для обеих таблиц, а how='left' задал левое объединение: используются ключи из левого датасета.

Из справочной таблицы cover_type_info добавлен один новый столбец — Avg_Temperature_C со средней температурой произрастания для каждого типа лесного покрова. Эти данные присоединились к каждой из 50 строк data1 в соответствии со значением Cover_Type в этой строке.

Пропущенных значений не обнаружено. Это объясняется тем, что:

  - Ключ Cover_Type присутствует в обеих таблицах
  - Все значения ключа из data1 есть в справочнике
