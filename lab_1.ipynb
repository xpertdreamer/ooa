**Задание 6**. График функции

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y = np.cos(x)
plt.plot(x, y)
plt.title("График функции y = cos(x)")
plt.grid(True)
plt.show()
```
