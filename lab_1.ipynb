## Задание 6. График функции

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 1000)
y = np.cos(x)
plt.plot(x, y)
plt.title("График функции y = cos(x)")
plt.grid(True)
plt.show()
```

#### Анализ

Первоначально здесь импортируем необходимые для работы библиотеки:
  
  - математическую библиотеку **numpy**
  - библиотеку для работы с визуализацией данных **matplotlib**

Создаем переменные, отражающие оси. При этом для оси x используем метод ``linspace`` библиотеки **numpy**, позволяющий создать массив равномерно расположенных чисел с указанным интервалом, а для оси y используем метод ``cos``.

С использованием методов **matplotlib** выводим получившуюся функцию на экран.
